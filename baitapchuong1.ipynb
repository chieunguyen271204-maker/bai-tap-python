```python
import math

# Kiểm tra với bộ cạnh (13, 14, 15)
a1, b1, c1 = 13, 14, 15
cv1 = a1 + b1 + c1
p1 = cv1 / 2
dt1 = math.sqrt(p1 * (p1 - a1) * (p1 - b1) * (p1 - c1))

print("--- Tam giác 1 (13, 14, 15) ---")
print("Chu vi:", cv1)
print("Diện tích:", dt1)

# Kiểm tra với bộ cạnh (3, 4, 5)
a2, b2, c2 = 3, 4, 5
cv2 = a2 + b2 + c2
p2 = cv2 / 2
dt2 = math.sqrt(p2 * (p2 - a2) * (p2 - b2) * (p2 - c2))

print("\n--- Tam giác 2 (3, 4, 5) ---")
print("Chu vi:", cv2)
print("Diện tích:", dt2)
```

```python
so_tien = 1_880_000
con_lai = so_tien

to_500 = con_lai // 500_000
con_lai %= 500_000

to_200 = con_lai // 200_000
con_lai %= 200_000

to_100 = con_lai // 100_000
con_lai %= 100_000

to_50 = con_lai // 50_000
con_lai %= 50_000

to_20 = con_lai // 20_000
con_lai %= 20_000

to_10 = con_lai // 10_000
con_lai %= 10_000

print("500.000 đồng:", to_500, "tờ")
print("200.000 đồng:", to_200, "tờ")
print("100.000 đồng:", to_100, "tờ")
print("50.000 đồng:", to_50, "tờ")
print("20.000 đồng:", to_20, "tờ")
print("10.000 đồng:", to_10, "tờ")
print("Số tiền còn lại:", con_lai)

tong = (
    to_500 * 500_000
    + to_200 * 200_000
    + to_100 * 100_000
    + to_50 * 50_000
    + to_20 * 20_000
    + to_10 * 10_000
)

print("Tổng tiền đã trả:", tong)
print("Kiểm tra:", tong == so_tien)
```

```python
n = 7539

# Tách chữ số
nghin = n // 1000
tram = (n % 1000) // 100
chuc = (n % 100) // 10
don_vi = n % 10

print("Số ban đầu:", n)
print("Hàng nghìn:", nghin)
print("Hàng trăm:", tram)
print("Hàng chục:", chuc)
print("Hàng đơn vị:", don_vi)

# Tổng các chữ số
tong_chu_so = nghin + tram + chuc + don_vi
print("Tổng các chữ số:", tong_chu_so)

# Số đảo ngược bằng toán học
so_dao_nguoc = don_vi * 1000 + chuc * 100 + tram * 10 + nghin
print("Số đảo ngược (toán học):", so_dao_nguoc)

# Kiểm tra lại bằng chuỗi theo đúng gợi ý đề bài
so_dao_nguoc_check = int(str(n)[::-1])
print("Kiểm tra bằng chuỗi:", so_dao_nguoc_check)
```
