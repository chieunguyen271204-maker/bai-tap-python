```python
#bai 4.1
import math

class HinhCau:
    def __init__(self, ban_kinh: float):
        if ban_kinh <= 0:
            raise ValueError("Bán kính hình cầu phải lớn hơn 0!")
        self.ban_kinh = ban_kinh

    def dien_tich(self) -> float:
        """Tính diện tích bề mặt hình cầu: 4 * pi * r^2"""
        return 4 * math.pi * (self.ban_kinh ** 2)

    def the_tich(self) -> float:
        """Tính thể tích hình cầu: (4/3) * pi * r^3"""
        return (4 / 3) * math.pi * (self.ban_kinh ** 3)

    def __str__(self) -> str:
        return f"Hình cầu (bán kính = {self.ban_kinh}, diện tích = {self.dien_tich():.3f}, thể tích = {self.the_tich():.3f})"
try:
    hc1 = HinhCau(2)
    print(hc1)
    print(f"Diện tích (kiểm tra riêng): {hc1.dien_tich():.3f}")
    print(f"Thể tích (kiểm tra riêng): {hc1.the_tich():.3f}")
except ValueError as e:
    print("Lỗi:", e)

print("-" * 40)
try:
    hc2 = HinhCau(-1)
    print(hc2)
except ValueError as e:
    print("Bắt được ngoại lệ thành công với r = -1:", e)
```

```python
#bai 4.3
class SoDuKhongDu(Exception):
    """Ngoại lệ khi số tiền rút vượt quá số dư hiện tại."""
    pass

class TaiKhoan:
    def __init__(self, chu_tai_khoan: str, so_du: float = 0):
        if so_du < 0:
            raise ValueError("Số dư ban đầu không được âm!")
        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
        self.lich_su = [f"Mở tài khoản với số dư: {so_du:,.0f}"]

    def nap(self, tien: float) -> None:
        if tien <= 0:
            raise ValueError("Số tiền nạp phải lớn hơn 0!")
        self.so_du += tien
        self.lich_su.append(f"+{tien:,.0f} (Số dư: {self.so_du:,.0f})")

    def rut(self, tien: float) -> None:
        if tien <= 0:
            raise ValueError("Số tiền rút phải lớn hơn 0!")
        if tien > self.so_du:
            raise SoDuKhongDu(f"Số dư ({self.so_du:,.0f}) không đủ để rút {tien:,.0f}!")
        self.so_du -= tien
        self.lich_su.append(f"-{tien:,.0f} (Số dư: {self.so_du:,.0f})")

    def __str__(self) -> str:
        return f"Tài khoản {self.chu_tai_khoan}: Số dư = {self.so_du:,.0f}"

tk = TaiKhoan("Nguyễn Văn A", 1_000_000)
tk.nap(500_000)
tk.rut(200_000)

try:
    tk.rut(2_000_000) 
except SoDuKhongDu as e:
    print("Lỗi giao dịch:", e)

print(tk)
print("Lịch sử giao dịch:", tk.lich_su)
assert tk.so_du == 1_300_000 and tk.so_du >= 0
```

```python
#bài 4.5
import math

class HinhHoc:
    def dien_tich(self) -> float:
        raise NotImplementedError("Lớp con phải cài đặt phương thức dien_tich()!")

    def __str__(self) -> str:
        return f"{type(self).__name__}: diện tích = {self.dien_tich():.4f}"

class HinhTron(HinhHoc):
    def __init__(self, r: float):
        if r <= 0:
            raise ValueError("Bán kính phải dương!")
        self.r = r

    def dien_tich(self) -> float:
        return math.pi * (self.r ** 2)

class HinhChuNhat(HinhHoc):
    def __init__(self, a: float, b: float):
        if a <= 0 or b <= 0:
            raise ValueError("Các cạnh hình chữ nhật phải dương!")
        self.a = a
        self.b = b

    def dien_tich(self) -> float:
        return self.a * self.b

class HinhTamGiac(HinhHoc):
    def __init__(self, a: float, b: float, c: float):
        if a <= 0 or b <= 0 or c <= 0 or (a + b <= c) or (a + c <= b) or (b + c <= a):
            raise ValueError("Ba cạnh không tạo thành tam giác hợp lệ!")
        self.a = a
        self.b = b
        self.c = c

    def dien_tich(self) -> float:
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))

danh_sach_hinh = [
    HinhTron(1),
    HinhChuNhat(3, 4),
    HinhTamGiac(3, 4, 5)
]

danh_sach_sap_xep = sorted(danh_sach_hinh, key=lambda h: h.dien_tich())

print("Các hình theo diện tích tăng dần:")
for hinh in danh_sach_sap_xep:
    print(" -", hinh)

tong_dien_tich = sum(h.dien_tich() for h in danh_sach_hinh)
print(f"Tổng diện tích: {tong_dien_tich:.4f}")
assert math.isclose(tong_dien_tich, math.pi + 18)
```

```python
import csv
import urllib.request
from collections import defaultdict
from pathlib import Path

duong_dan = Path("data/goldprice.csv")
if not duong_dan.exists():
    duong_dan.parent.mkdir(parents=True, exist_ok=True)
    url = "https://raw.githubusercontent.com/datasets/gold-prices/master/data/monthly.csv"
    urllib.request.urlretrieve(url, duong_dan)

gia_theo_nam = defaultdict(list)

with open(duong_dan, mode="r", encoding="utf-8") as f:
    for dong in csv.DictReader(f):
        try:
            nam = int(dong["Date"][:4])
            gia = float(dong["Price"])
            gia_theo_nam[nam].append(gia)
        except (ValueError, KeyError, TypeError):
            continue

trung_binh_nam = {
    nam: sum(ds_gia) / len(ds_gia)
    for nam, ds_gia in gia_theo_nam.items()
    if len(ds_gia) == 12
}

cac_nam = sorted(trung_binh_nam.keys())
muc_tang = []

for i in range(1, len(cac_nam)):
    nam_truoc, nam_nay = cac_nam[i - 1], cac_nam[i]
    if nam_nay == nam_truoc + 1:
        ty_le = (trung_binh_nam[nam_nay] / trung_binh_nam[nam_truoc] - 1) * 100
        muc_tang.append((nam_nay, ty_le))

muc_tang_giam_dan = sorted(muc_tang, key=lambda x: x[1], reverse=True)

print("Ba năm giá vàng tăng mạnh nhất:")
for nam, phan_tram in muc_tang_giam_dan[:3]:
    print(f" - Năm {nam}: +{phan_tram:.2f}%")

nam_giam, phan_tram_giam = muc_tang_giam_dan[-1]
print(f"\nNăm giá vàng giảm mạnh nhất:")
print(f" - Năm {nam_giam}: {phan_tram_giam:.2f}%")
```
